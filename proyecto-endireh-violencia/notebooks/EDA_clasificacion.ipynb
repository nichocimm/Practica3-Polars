## Clasificación de Variables

<center>

| Variable | Tipo de dato | Justificación |
| :---: | :---: | :---|
| edad_primer_union | Cuantitativa | Representa el año en el que la mujer tuvo al unirse por primera vez. Es numérico y entero |
| num_hijos | Cuantitativa | Representa cuantos hijos tiene la mujer. Es un número entero y numérico. |
| nivel_escolaridad  | Cualítativa Ordinal | Representa el nivel de escolaridad que tiene la mujer, en una escala de 'A1', 'A2', 'B1', 'B2', 'C1', 'C2'  con una relación de orden entre letras y números. |
| nom_entidad | Cualítativa Nominal | Representa la entidad federativa en donde reside la mujer. |
| nom_municipio | Cualítativa Nominal | Representa el municipio de la entidad federativa donde reside la mujer. |
| estado_civil_desc | Cualítativa Nominal | Representa el estado civil de la mujer |
| estrato_socioeconomico | Cualítativa Ordinal | Aunque se represente numericamente, se trata de diferentes estados economicos con orden entre si |
| pareja_trabaja_desc | Cualítativa Nominal | Describe si la pareja trabaja o no |
| ingreso_pareja | Cuantitativa  | Cantidad númerica que recibe la pareja de la mujer |
| dinero_propio_desc | Cualítativa Nominal | Describe si tiene dinero propio o no, la mujer |
| apoyo_gobierno_desc | Cualítativa Nominal | Describe si tiene apoyo del gobierno o no, la mujer |
| tiene_ahorros_desc | Cualítativa Nominal | Describe si la mujer tiene ahorros o no |
| propietaria_vivienda_desc | Cualítativa Nominal | Describe si la mujer es propietaria de la vivienda |
| sufrio_violencia_pareja | Cualítativa Nominal | Describe si la mujer ha sufrido violencia de pareja |
| factor_expansion | Cuantitativa  | Es un peso estadístico calculado por el INEGI que es númerico. Indica a cuántas mujeres reales representa cada fila de la muestra. |
| autonomia_financiera | Cualítativa Ordinal | Representa con los identificadores de 'baja','media','alta', la autonomía financiera de la mujer |
| bloque_educativo | Cualítativa Ordinal | Representa en que bloque educativo se encuentra la mujer |
